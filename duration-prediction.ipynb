```python
import pandas as pd

url = "https://d37ci6vzurychx.cloudfront.net/trip-data/green_tripdata_2021-01.parquet"

df = pd.read_parquet(url)

df.head()
```

```python
print("Rows and columns:", df.shape)
df.info()
```

```python
df.info()
```

```python
print(df.shape)
```

```python
df["duration"] = (
    df["lpep_dropoff_datetime"] - df["lpep_pickup_datetime"]
).dt.total_seconds() / 60

df[[
    "lpep_pickup_datetime",
    "lpep_dropoff_datetime",
    "duration"
]].head()
```

```python
df["duration"].describe()
```

```python
df_clean = df.loc[
    (df["duration"] >= 1) & (df["duration"] <= 60)
].copy()

print("Original trips:", len(df))
print("Trips kept:", len(df_clean))
print("Trips removed:", len(df) - len(df_clean))

df_clean["duration"].describe()
```
